```python
values_tempLP = [1e-14,1e-13,1e-12]
for iv, v in enumerate(values_tempLP):
    params_aux = params_env.copy()
    params_aux['tempLP'] = v
    env_aux = utils.environment_exp(**params_aux)

    S, Y, Z, X, nu, wealth, perf, rew, venue_wealth, A_a_hist, A_b_hist, A_W_hist, A_B_hist, rew_drift_hist = env_aux.simulate_venue(nsims=nsims, strag_LP=strag_LP, strag_vn=strag_vn, verbose = True)
    qtl_nu = np.zeros((len(env.timesteps),3))

    dt = env.timesteps[1] - env.timesteps[0]
    qtl_dQ = np.zeros((len(env.timesteps),3))
    dQ = dt * np.cumsum(nu[:,:], axis = 0)

    for it, t in enumerate(env.timesteps):
        qtl_nu[it, :] = np.quantile(nu[it,:],[0.05,0.5,0.95],axis=0)
        qtl_dQ[it, :] = np.quantile(dQ[it,:],[0.05,0.5,0.95],axis=0)
        
    fig, axs = plt.subplots(1, 1, sharex=True, figsize=(6,5))
    plt.plot(env.timesteps, dt * np.cumsum(nu[:,nsim],axis=0), label=r'$\nu$', color ='tab:red')
    plt.fill_between(env.timesteps,qtl_dQ[:,0],qtl_dQ[:,-1], color='tab:blue', alpha=0.25)
    plt.xlabel(r'$t$', fontsize = 18)
    plt.ylabel(r'$\mathrm{change}$~$\mathrm{in}$~$\mathrm{liquidity}$~$\mathrm{provided}$', fontsize = 22)
    plt.hlines(y = 0, xmin = 0, xmax = 1, linestyle = '--', alpha = 0.4, color='tab:red')
    plt.ylim([-30,30])
    plt.tight_layout()
    plt.savefig(f'figures/baseline-tempLP{iv}-sim-cumnu-paths-{risk_type}.pdf', format="pdf", dpi=dpi)
    plt.show()
    
    
    fig, axs = plt.subplots(1, 1, sharex=True, figsize=(6,5))
    plt.plot(env.timesteps, nu[:,nsim], label=r'$\nu$', color ='tab:red')
    plt.fill_between(env.timesteps,qtl_nu[:,0],qtl_nu[:,-1], color='tab:blue', alpha=0.25)
    plt.xlabel(r'$t$', fontsize = 18)
    plt.ylabel(r'$\nu^*_t$', fontsize = 22)
    plt.hlines(y = 0, xmin = 0, xmax = 1, linestyle = '--', alpha = 0.4, color='tab:red')
    plt.ylim([-1800,1800])
    plt.tight_layout()
    plt.savefig(f'figures/baseline-tempLP{iv}-sim-nu-paths-{risk_type}.pdf', format="pdf", dpi=dpi)
    plt.show()
```
